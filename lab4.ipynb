```python
import time
import timeit
import tracemalloc
import numpy as np

N = 1_000_000

py_list = list(range(N))
np_arr = np.arange(N)

def loop_double(values):
    result = []
    for x in values:
        result.append(x * 2)
    return result

def numpy_double(values):
    return values * 2


for label, func, data in [
    ("loop", loop_double, py_list),
    ("numpy", numpy_double, np_arr)
]:
    start = time.perf_counter()
    func(data)
    elapsed = time.perf_counter() - start
    print(label, round(elapsed, 5), "seconds")
```

```python
import timeit
import numpy as np

def loop_double(values):
    result = []
    for x in values:
        result.append(x * 2)
    return result

def numpy_double(values):
    return values * 2

rows = []

for N in [1_000, 100_000, 1_000_000]:
    py_list = list(range(N))
    np_arr = np.arange(N)

    # Use more repetitions for small arrays
    n = 100 if N == 1_000 else 3

    t_loop = min(timeit.repeat(
        lambda: loop_double(py_list), number=n, repeat=3
    )) / n

    t_np = min(timeit.repeat(
        lambda: numpy_double(np_arr), number=n, repeat=3
    )) / n

    ratio = t_loop / t_np
    rows.append((N, t_loop, t_np, ratio))

print("N | Loop seconds | NumPy seconds | Ratio")
for row in rows:
    print(row[0], round(row[1], 6),
          round(row[2], 6), round(row[3], 2))
```

Size Sweep Conclusion

I compared loop-based and NumPy vectorized operations for different array sizes. I used the measured results to compare their execution times and speed ratios. NumPy was faster in my measurements. The speed ratio changed with array size, as shown in my results.

```python
n = 20

t_loop = timeit.timeit(
    'loop_double(py_list)',
    globals=globals(),
    number=n
) / n

t_np = timeit.timeit(
    'numpy_double(np_arr)',
    globals=globals(),
    number=n
) / n

print("Loop average:", round(t_loop, 5), "seconds")
print("NumPy average:", round(t_np, 5), "seconds")
print("Speed ratio:", round(t_loop / t_np, 2))
```

```python
N = 1_000_000
py_list = list(range(N))
np_arr = np.arange(N)

n = 3

t_loop = min(timeit.repeat(
    lambda: loop_double(py_list), number=n, repeat=5
)) / n

t_np = min(timeit.repeat(
    lambda: numpy_double(np_arr), number=n, repeat=5
)) / n

print("Best loop time:", t_loop)
print("Best NumPy time:", t_np)
print("Speed ratio:", t_loop / t_np)
```
